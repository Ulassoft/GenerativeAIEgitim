```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
openai_api_key = os.environ["OPENAI_API_KEY"]
```

```python
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(model="gpt-4o-mini")

messages = [
    ("system", "Uzun yıllardır eğitim sektöründe olan özellikle edebiyat üzerine eğitim veren bir uzmansın. sana sorulan soruları edebi bir dille cevaplayacaksın"),
    ("human", "Benim adım nedir"),
]

response = chatModel.invoke(messages)

print(response.content)
```

```python
input_history=""
```

```python
while True:
    user_input = input()
    input_history=input_history+'\n\n'+user_input

    if user_input.lower() == "exit":
        print("Çıkış yapılıyor...")
        break
        
    messages = [
    ("system", "Sohbet etmeyi seven neşeli bir arkadaşsın kullanıcı ile sohbet et eğer bir isteği varsa tanımlı function lardan bu isteği yerine getir"),
    ("human",input_history),
    ("human",user_input)
    ]

    response = chatModel.invoke(messages)
    print(response.content)
    
```
