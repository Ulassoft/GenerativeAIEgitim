```python
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(model="gpt-4o-mini")
```

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader("./LOR.txt", encoding="utf-8")

loaded_data = loader.load()
```

```python
loaded_data
```

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=1000,
    chunk_overlap=100,
    length_function=len,
    is_separator_regex=False,
)
```

```python
texts = text_splitter.create_documents([loaded_data[0].page_content])
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings_model = OpenAIEmbeddings()
```

```python
kaplan_embedding = embeddings_model.embed_query("kaplan")

# Embedding'i yazdır
print(kaplan_embedding)
```

```python
from langchain_chroma import Chroma
```

```python
vector_db = Chroma.from_documents(texts, OpenAIEmbeddings())
```

```python
question = "who made the one ring"

response = vector_db.similarity_search(question)

print(response[0].page_content)
```

```python
retriever = vector_db.as_retriever()
```

```python
response = retriever.invoke("who made the one ring")
```

```python
response
```

```python
retriever = vector_db.as_retriever(search_kwargs={"k": 1})
```

```python
response = retriever.invoke("who made the one ring")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
template = """Sana verdiğim  bilgilere dayanarak soruyu kısaca cevapla:
bilgi:
{context}

Soru: {question}
"""
```

```python
prompt = ChatPromptTemplate.from_template(template)
```

```python
def format_docs(docs):
    return "\n\n".join([d.page_content for d in docs])
```

```python
chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | chatModel
    | StrOutputParser()
)
```

```python
response = chain.invoke("who made the one ring")
```

```python
response
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults
```

```python
search = TavilySearchResults(max_results=2)
search_sonuc=search.invoke("Who made Harry Potter's wand?")
```

```python
search_sonuc[0]['content']
```

```python
def sonuclarigetir(question):
    cevap = vector_db.similarity_search_with_score(question,k=2)
    soncevap=""
    if cevap[0][1]<0.25 or cevap[1][1]<0.25 :
        soncevap=cevap[0][0].page_content+"\n\n"+cevap[1][0].page_content
    else:
        search_sonuc=search.invoke(question)
        soncevap=search_sonuc[0]['content']+"\n\n"+search_sonuc[1]['content']        
    return(soncevap)  
```

```python
sonuclarigetir("Who made Harry Potter's wand?")
```

```python
chain2 = (
    {"context": sonuclarigetir, "question": RunnablePassthrough()}
    | prompt
    | chatModel
    | StrOutputParser()
)
```

```python
response2 = chain2.invoke("who made the one ring")
response2
```

```python
response3 = chain2.invoke("Who made Harry Potter's wand?")
response3
```
