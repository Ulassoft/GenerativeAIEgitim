```python
import pandas as pd
import mediapipe as mp
import cv2
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

import numpy as np
```

```python
def calculate_angle(a,b,c):
    radians = np.arctan2(c.y-b.y, c.x-b.x) - np.arctan2(a.y-b.y, a.x-b.x)
    angle = np.abs(radians * 180.0 / np.pi)

    if angle > 180.0:
        angle = 360 - angle

    return angle
```

```python
# 4. Make Predictions with a Test Image
# Initialize the Mediapipe Pose module
mp_pose = mp.solutions.pose
pose = mp_pose.Pose(static_image_mode=True, model_complexity=1, enable_segmentation=False)
```

```python
# Load a test image (test.jpg)
test_image_path = "testimages/test2.jpg"  # Specify the path to your test image here
test_image = cv2.imread(test_image_path)

# Convert the image to RGB (Mediapipe works in RGB format)
test_image_rgb = cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB)

# Perform pose estimation with the test image
results = pose.process(test_image_rgb)

# If landmarks are detected, make predictions
if results.pose_landmarks:
    # Convert landmark data into a single row

    ru=calculate_angle(results.pose_landmarks.landmark[12],results.pose_landmarks.landmark[24],results.pose_landmarks.landmark[26])
    rd=calculate_angle(results.pose_landmarks.landmark[24],results.pose_landmarks.landmark[26],results.pose_landmarks.landmark[28])
    lu=calculate_angle(results.pose_landmarks.landmark[11],results.pose_landmarks.landmark[23],results.pose_landmarks.landmark[25])
    ld=calculate_angle(results.pose_landmarks.landmark[23],results.pose_landmarks.landmark[25],results.pose_landmarks.landmark[27])

    mypose=""
    if  ru<110  and rd<110 and lu >135  and ld<110 :
        mypose="rlunges"
    elif  lu<110  and ld<110 and ru >135  and rd<110 :
        mypose="llunges"
    elif ru>135 and  rd>135 and lu>135 and ld>135 :
        mypose="zero"
    else: 
        mypose="other"

        
    print("ru",ru)
    print("rd",rd)
    print("lu",lu)
    print("ld",ld)
    print(mypose)
```

```python

# Close the Pose module
pose.close()
```
