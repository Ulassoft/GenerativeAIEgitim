```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
```

```python
hugging_face_token = os.environ["HUGGINGFACE_TOKEN"]
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilgpt2", token=hugging_face_token)
model = AutoModelForCausalLM.from_pretrained("distilbert/distilgpt2", token=hugging_face_token)
```

```python
input_text = "who is newtons father"
input_ids = tokenizer.encode(input_text, return_tensors="pt")
output_ids = model.generate(input_ids, max_length=50)
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

#print("Üretilen metin:")
print(output_text)
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", token=hugging_face_token)
model = AutoModelForCausalLM.from_pretrained("deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", token=hugging_face_token)
```

```python
input_text = "who is newtons father"
input_ids = tokenizer.encode(input_text, return_tensors="pt")
output_ids = model.generate(input_ids, max_length=50)
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

#print("Üretilen metin:")
print(output_text)
```

```python
# Girdi hazırlanıyor
input_text = "who is newtons father"
inputs = tokenizer(input_text, return_tensors="pt", padding=True)

# Pad token id tanımlanmadıysa, model'e bunu belirt
if model.config.pad_token_id is None:
    model.config.pad_token_id = tokenizer.eos_token_id

# Metin üretiliyor
output_ids = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_length=50,
    pad_token_id=model.config.pad_token_id  # Güvenli açık uçlu üretim
)

# Çıktı çözülüyor
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

# Sonuç
print(output_text)
```
