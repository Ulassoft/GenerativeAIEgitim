```python
import openai
import os

# OpenAI API key'in
openai.api_key = os.environ["OPENAI_API_KEY"]

# Yeni API'ye uygun dosya yükleme
with open("hayalgucu_finetune_data.jsonl", "rb") as f:
    file_response = openai.files.create(file=f, purpose="fine-tune")

print("Dosya yüklendi. ID:", file_response.id)
```

```python
fine_tune_response = openai.fine_tuning.jobs.create(
    training_file=file_response.id,
    model="gpt-4o-mini-2024-07-18"
)

print("Fine-tuning job başlatıldı. ID:", fine_tune_response.id)
```

```python
job_id = fine_tune_response.id  # senin job ID

status = openai.fine_tuning.jobs.retrieve(job_id)
print("Durum:", status.status)
```

```python
job_id = fine_tune_response.id  # veya doğrudan ID'yi yazabilirsin

status = openai.fine_tuning.jobs.retrieve(job_id)

model_name = status.fine_tuned_model
print("Model ismi:", model_name)
```
