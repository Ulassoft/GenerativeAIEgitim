```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
```

```python
hugging_face_token = os.environ["HUGGINGFACE_TOKEN"]
```

```python
hugging_face_token
```

```python
import warnings
```

```python
warnings.filterwarnings('ignore')
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("ulascelenk/deepseekColab", token=hugging_face_token)
model = AutoModelForCausalLM.from_pretrained("ulascelenk/deepseekColab", token=hugging_face_token)
```

```python
# Girdi hazırlanıyor
input_text = "### User: Where is Hayalgucu Patisserie located?\n### Assistant:"
inputs = tokenizer(input_text, return_tensors="pt", padding=True)

# Pad token id tanımlanmadıysa, model'e bunu belirt
if model.config.pad_token_id is None:
    model.config.pad_token_id = tokenizer.eos_token_id

# Metin üretiliyor
output_ids = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_length=50,
    pad_token_id=model.config.pad_token_id  # Güvenli açık uçlu üretim
)

# Çıktı çözülüyor
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=False)

# Eğer model yeni bir user prompt'u başlattıysa, oradan kes
cleaned_output = output_text.split("### User:")[1].strip()

print("\n🧠 Model Yanıtı:\n", cleaned_output)

```
