```python
import requests

api_key = "XXXXX"
headers = {"accept": "application/json", "x-api-key": api_key}
template_list_url = "https://api.heygen.com/v2/templates"

response = requests.get(template_list_url, headers=headers)
templates = response.json().get("data", {}).get("templates", [])
if templates:
    template_id = templates[0]["template_id"]  # İlk şablonun ID'sini kullanın
    print(f"Template ID: {template_id}")
else:
    print("No templates found.")
```

```python
template_url = f"https://api.heygen.com/v2/template/{template_id}"
response = requests.get(template_url, headers=headers)
print(response.json())
variables = response.json().get("data", {}).get("variables", {})
print(response.json())
print(variables)
```

```python
variables_values = {
        "sozler": {
            "name": "sozler",
            "type": "text",
            "properties": {
                "content": "Hello, how are you? It’s a pleasure to meet you. First of all, I want to thank you for joining our AI lessons. I hope you enjoy the training sessions and the subjects we’ll be exploring."
            }
        }
}
```

```python
generate_url = f"https://api.heygen.com/v2/template/{template_id}/generate"
payload = {
    "caption": False,
    "title": "Ilk video",
    "locale": "en-US",
    "dimension": {
        "width": 1280,
        "height": 720
          },
    "variables": variables_values
}
headers["Content-Type"] = "application/json"

response = requests.post(generate_url, headers=headers, json=payload)
if response.status_code == 200:
    video_id = response.json().get("data", {}).get("video_id")
    print(f"Video ID: {video_id}")
else:
    print(f"Error: {response.status_code} - {response.text}")
```

```python
import time

video_status_url = f"https://api.heygen.com/v1/video_status.get?video_id={video_id}"

while True:
    response = requests.get(video_status_url, headers=headers)
    status = response.json().get("data", {}).get("status")
    if status == "completed":
        video_url = response.json().get("data", {}).get("video_url")
        video_content = requests.get(video_url).content
        with open("generated_video.mp4", "wb") as video_file:
            video_file.write(video_content)
        print("Video downloaded successfully.")
        break
    elif status in ["processing", "pending"]:
        print("Video is still processing. Checking status...")
        time.sleep(5)
    elif status == "failed":
        error = response.json().get("data", {}).get("error")
        print(f"Video generation failed. Error: {error}")
        break
```
